```python
#Penjumlahan
print("Hasil Penjumlahan 13+2 =", 13+2)
mangga = 15
stoberi = 7
buah = mangga + stoberi
print("Sisa Buah", buah)

#Pengurangan
hutang = 10000
bayar = 6700
sisahutang = hutang - bayar
print("Sisa hutang anda adalah ", sisahutang)

#Perkalian
p = 6
l = 7
L = p * l
print("Hasil luas: ", L)

#Pembagian
nasikotak = 15
peserta = 5
dapat = nasikotak / peserta
print("Hasil nasi kotak yang didapatkan per peserta adalah", dapat)

#Sisa Bagi
x = 18
y = 14
hasil = x % y
print("Sisa bagi dari bilangan", x, "dan", y, "adalah", hasil)

#Pangkat
w = 123
z = 4
hasilpangkat = w ** z
print("Hasil pangkat: ", hasilpangkat)

#Pembagian Bilangan Bulat
b = 10
c = 9
print("Hasil pembagian 10/9 adalah ", b/c)
print("Hasil dari pembulatannya adalah", 10//9)

```
