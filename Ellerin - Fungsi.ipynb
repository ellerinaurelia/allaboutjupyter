```python
def jumlah(bil1, bil2):
    jumbil=bil1+bil2
    return jumbil
```

```python
a=11
b=55
hitung=jumlah(a,b)
print("Hasil Penjumlahan", a, "+", b, "adalah" , hitung)
```
