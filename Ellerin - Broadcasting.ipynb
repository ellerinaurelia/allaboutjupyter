Apakah proses pertambahan bisa dilakukan untuk objek NumPy dengan dimensi 2x3 dan 3x2 ?

Tidak bisa, Operasi aritmatika dasar (seperti +, -, *, /) antar array di NumPy menggunakan aturan yang disebut Broadcasting. Agar dua matriks bisa ditambahkan, dimensinya harus sama persis, atau memenuhi aturan broadcasting (dimensi harus cocok dari ujung kanan/belakang, dan jika berbeda, salah satunya harus bernilai 1).

Karena matriks (2, 3) dan (3, 2) memiliki ukuran yang saling terbalik dan tidak ada dimensi yang bernilai 1 untuk di-broadcast, NumPy tidak akan tahu elemen mana yang harus ditambahkan dengan mana. Jika kamu mencobanya (a + b), Python akan mengeluarkan pesan error: ValueError: operands could not be broadcast together.
